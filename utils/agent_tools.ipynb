```python
# Imports and Variable Set Up
import logging

logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
logger = logging.getLogger(__name__)
```

```python
def execute_in_databricks(sql_code):
    """
    Presents proposed SQL code to the user for review and executes
    it in Databricks upon confirmation.

    Args:
        sql_code (str): The SQL code proposed by the agent

    Returns:
        str: Execution result or cancellation message
    """
    print("\n" + "="*50)
    print("⚠️  PROPOSED SQL — REVIEW BEFORE EXECUTING:")
    print("="*50)
    print(sql_code)
    print("="*50 + "\n")

    confirmation = input("Execute this SQL? (yes/no): ").strip().lower()

    if confirmation == "yes":
        try:
            result = spark.sql(sql_code)
            rows = result.collect()
            if rows:
                output = "\n".join([str(row) for row in rows])
                logger.info(f"✅ SQL executed successfully. Rows returned: {len(rows)}")
                return f"SQL executed successfully and committed to Databricks. Results:\n{output}"
            else:
                logger.info("✅ SQL executed successfully. No rows returned.")
                return "SQL executed successfully and committed to Databricks. No further action required."
        except Exception as e:
            logger.error(f"❌ SQL execution failed: {e}")
            return f"SQL execution failed: {str(e)}"
    else:
        logger.info("🚫 Execution cancelled by user.")
        return "Execution was cancelled by the user. No SQL was run."
```
